💡 **Dica:** `coluna_x.cov(coluna_y)` e `coluna_x.corr(coluna_y)`. A correlação também é `covariância / (desvio_x × desvio_y)`.

```python
horas = dados["horas_estudo_semana"]

cov_horas_nota = horas.____(notas)
r_horas_nota   = horas.____(notas)

# Conferindo "na mão": covariância dividida pelo produto dos desvios padrão
r_manual = cov_horas_nota / (horas.std() * notas.std())

print(f"Covariância = {cov_horas_nota:.2f}")
print(f"Correlação (.corr())  = {r_horas_nota:.2f}")
print(f"Correlação (na mão)   = {r_manual:.2f}")
```

### Exercício 5.2: Dispersão com interpretação automática
A função abaixo desenha o gráfico, calcula o `r` e escreve a interpretação. **Ela já está pronta.** Use-a para três pares de variáveis.

```python
def interpretar_r(r):
    if abs(r) < 0.3:
        return "fraca ou inexistente"
    forca = "moderada" if abs(r) < 0.7 else "forte"
    direcao = "positiva" if r > 0 else "negativa"
    return f"{forca} {direcao}"

def desenhar_dispersao(coluna_x, coluna_y):
    r = dados[coluna_x].corr(dados[coluna_y])
    sns.regplot(data=dados, x=coluna_x, y=coluna_y, ci=None,
                scatter_kws={"alpha": 0.4}, line_kws={"color": "red"})
    plt.title(f"{coluna_x} x {coluna_y}\nr = {r:.2f} ({interpretar_r(r)})")
    plt.show()
    return r
```

```python
renda = dados["renda_familiar_mensal"]
rq1, rq3 = renda.quantile(0.25), renda.quantile(0.75)
limite_renda = rq3 + 1.5 * (rq3 - rq1)

dados["renda_e_outlier"] = renda ____ limite_renda        # True para quem está acima do limite
n_outliers_renda = dados["renda_e_outlier"].sum()

sem_outliers = dados[____dados["renda_e_outlier"]]        # novo DataFrame, sem os outliers

r_renda_com = dados["renda_familiar_mensal"].corr(dados["nota_final"])
r_renda_sem = sem_outliers["renda_familiar_mensal"].corr(sem_outliers["nota_final"])

print(f"Limite superior da renda: R$ {limite_renda:,.2f}")
print(f"Outliers: {n_outliers_renda} alunos ({n_outliers_renda / len(dados) * 100:.1f}%)")
print(f"r COM outliers: {r_renda_com:.2f} | r SEM outliers: {r_renda_sem:.2f}")
```
